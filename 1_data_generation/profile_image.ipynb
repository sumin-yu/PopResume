```python
import os
import json
import pandas as pd

profile_img_path = "/path/to/CausalFair/resume_images_new/output_0224_fix_race_AIAN"

img_list = os.listdir(profile_img_path)
img_list = [img for img in img_list if img.endswith(".jpg") or img.endswith(".png")]
img_list = sorted(img_list)
print(f"Total number of profile images: {len(img_list)}")
meta_data_path = "/path/to/CausalFair/Resume/job-distribution/processed_job_data_0102_with_exp_pred_with_names.csv"
meta_data = pd.read_csv(meta_data_path)
print(meta_data["race"].value_counts())
aian_id_list = meta_data[meta_data["race"] == "AIAN"]["id"].tolist()
show_num = 10
aian_id_list = aian_id_list[:show_num]
print(f"AIAN ID List: {aian_id_list}")

import matplotlib.pyplot as plt
for img in img_list:
    img_id = int(img.split(".")[0].split("_")[0])
    if img_id in aian_id_list:
        img_path = os.path.join(profile_img_path, img)
        img_data = plt.imread(img_path)
        plt.imshow(img_data)
        plt.title(f"ID: {img_id} - Race: AIAN")
        plt.axis("off")
        plt.show()
```

```python
meta_data_aian = meta_data[meta_data["race"] == "AIAN"]
# meta_data_aian['job'].value_counts().plot(kind='bar')
total_job_count = meta_data["job"].value_counts()
aian_job_count = meta_data_aian["job"].value_counts()
job_categories = total_job_count.index
total_counts = total_job_count.values
aian_counts = aian_job_count.reindex(job_categories, fill_value=0).values
for job, total, aian in zip(job_categories, total_counts, aian_counts):
    print(f"Job: {job}, Total Count: {total}, AIAN Count: {aian} ({(aian/total)*100:.2f}%)")
```
